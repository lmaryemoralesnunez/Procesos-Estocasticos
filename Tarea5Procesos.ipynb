EJERCICIO 3

```python
import numpy as np
from math import exp, factorial, sqrt

# matriz R
R = np.array([
    [0,2,3,0],
    [4,0,2,0],
    [0,2,0,2],
    [1,0,3,0]
],dtype=float)

# r_i
ri = np.sum(R,axis=1)

# r=max(r_i)
r = np.max(ri)

print("r =",r)

# matriz P sombrero
P_hat = np.zeros_like(R)

for i in range(4):
    for j in range(4):
        if i==j:
            P_hat[i,j] = 1-ri[i]/r
        else:
            P_hat[i,j] = R[i,j]/r

print(P_hat)
```

Función para calcular P(t)


```python
def P_t(t):

    M = int(max(r*t + 5*np.sqrt(r*t),20))

    n = len(P_hat)

    P = np.zeros((n,n))

    for k in range(M+1):

        coef = exp(-r*t)*(r*t)**k/factorial(k)

        P += coef*np.linalg.matrix_power(P_hat,k)

    return P
```

Calculando

```python
print("P(0.5)")
print(P_t(0.5))

print()

print("P(1)")
print(P_t(1))

print()

print("P(5)")
print(P_t(5))
```

Chapman-Kolmogorov

```python
P05 = P_t(0.5)
P1 = P_t(1)

print(P1)

print()

print(P05 @ P05)
```

Algoritmo de Uniformización para calcular P(t)

```python
import numpy as np

def uniformizacion(R, t, epsilon=1e-5):

    # Paso 1: calcular r
    r = np.max(np.sum(R, axis=1))

    # Paso 2: construir P_hat
    n = R.shape[0]
    P_hat = np.zeros((n,n))

    for i in range(n):
        ri = np.sum(R[i,:])

        for j in range(n):

            if i == j:
                P_hat[i,j] = 1 - ri/r
            else:
                P_hat[i,j] = R[i,j]/r

    # Paso 3: inicialización
    A = P_hat.copy()

    B = np.exp(-r*t)*np.eye(n)

    c = np.exp(-r*t)

    suma = c

    k = 1

    # Paso 4: ciclo principal
    while suma < 1 - epsilon:

        c = c*(r*t)/k

        B = B + c*A

        A = A @ P_hat

        suma += c

        k += 1

    return B, k-1
```

```python
R = np.array([
    [0,2,3,0],
    [4,0,2,0],
    [0,2,0,2],
    [1,0,3,0]
], dtype=float)
```

```python
P05, M05 = uniformizacion(R, 0.5)

print("M utilizado =", M05)
print(P05)
```

```python
P1, M1 = uniformizacion(R, 1)

print("M utilizado =", M1)
print(P1)
```

```python
P5, M5 = uniformizacion(R, 5)

print("M utilizado =", M5)
print(P5)
```

Verificar Chapman-Kolmogorov

```python
izquierda = P1

derecha = P05 @ P05

print("P(1):")
print(izquierda)

print("\nP(0.5)P(0.5):")
print(derecha)

print("\nError máximo:")
print(np.max(np.abs(izquierda - derecha)))
```
