```python
#Laura Maryem Morales Nuñez
# SYMPY

from sympy import Matrix

# 1. Crear matrices
A = Matrix([[1, 2],
            [3, 4]])

B = Matrix([[0, 1],
            [1, 0]])

print("Matriz A:")
print(A)

print("\nMatriz B:")
print(B)


# 2. Obtener entradas, filas y columnas
entrada = A[0, 1]      # Fila 0, columna 1
fila = A.row(0)        # Primera fila
columna = A.col(1)     # Segunda columna

print("\nEntrada A[0,1]:", entrada)
print("Primera fila:", fila)
print("Segunda columna:", columna)


# 3. Multiplicar matrices
C = A * B
print("\nProducto A*B:")
print(C)


# 4. Elevar matriz a una potencia
A2 = A**2
print("\nA^2:")
print(A2)


# 5. Multiplicar vector por matriz
v = Matrix([1, 0])   # Vector columna
resultado = A * v

print("\nVector v:")
print(v)

print("A * v:")
print(resultado)


# 6. Valores y vectores propios
valores_propios = A.eigenvals()
vectores_propios = A.eigenvects()

print("\nValores propios:")
print(valores_propios)

print("\nVectores propios:")
print(vectores_propios)

# NETWORKX - Diagrama transición

import networkx as nx
import matplotlib.pyplot as plt

# Crear grafo dirigido
G = nx.DiGraph()

# Agregar nodos
G.add_nodes_from(['A', 'B'])

# Agregar transiciones con probabilidades
G.add_edge('A', 'A', weight=0.8)
G.add_edge('A', 'B', weight=0.2)
G.add_edge('B', 'A', weight=0.5)
G.add_edge('B', 'B', weight=0.5)

# Posiciones de nodos
pos = {'A': (1,0), 'B': (-1,0)}

plt.figure(figsize=(6,4))

# Dibujar nodos
nx.draw_networkx_nodes(G, pos, node_size=2000)

# Dibujar etiquetas de nodos
nx.draw_networkx_labels(G, pos, font_size=12)

# Dibujar aristas curvas
nx.draw_networkx_edges(G, pos, connectionstyle='arc3,rad=0.2', arrows=True)

# Etiquetas de probabilidades
edge_labels = nx.get_edge_attributes(G,'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.axis('off')
plt.show()
```
