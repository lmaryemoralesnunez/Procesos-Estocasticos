# Laberinto

Modelo del juego

Los estados del tablero son:

$
1,2,3,...,20
$

donde el estado \(20\) es absorbente, ya que representa la meta del juego.

En cada turno se lanza un dado justo de 6 caras.

Además, el tablero contiene serpientes y escaleras:

- Si el jugador cae en \(17\), baja a \(10\).
- Si el jugador cae en \(13\), baja a \(4\).
- Si el jugador cae en \(3\), sube a \(11\).
- Si el jugador cae en \(15\), sube a \(19\).


Parte analítica
Sea:

$
E_i
$

el número esperado de tiradas necesarias para llegar al estado final \(20\), comenzando desde el estado \(i\).

Como el estado \(20\) es absorbente:

$
E_{20}=0
$

Para cualquier otro estado se cumple:

$
E_i=1+\frac{1}{6}\sum_{k=1}^{6} E_{f(i,k)}
$

donde:

- \(f(i,k)\) representa la casilla a la que llega el jugador desde el estado \(i\) al obtener \(k\) en el dado.
- La función incluye automáticamente las serpientes y escaleras.

La suma representa el promedio sobre las seis posibles tiradas del dado.

```python
import numpy as np

N = 20

# serpientes y escaleras
especiales = {
    3:11,
    15:19,
    13:4,
    17:10
}

# matriz del sistema
A = np.zeros((N,N))
b = np.zeros(N)

# estado absorbente
A[19,19] = 1
b[19] = 0

for i in range(19):

    A[i,i] = 1
    b[i] = 1

    for dado in range(1,7):

        j = i + 1 + dado

        if j > 20:
            j = 20

        if j in especiales:
            j = especiales[j]

        A[i,j-1] -= 1/6

# resolver sistema
E = np.linalg.solve(A,b)

print("Esperanza desde cada estado:\n")
for i,val in enumerate(E, start=1):
    print(f"E[{i}] = {val:.4f}")

print("\nNúmero promedio de tiradas desde el inicio:")
print(E[0])
```

Código simulación

```python
import numpy as np

especiales = {
    3:11,
    15:19,
    13:4,
    17:10
}

def jugar():

    posicion = 1
    tiros = 0

    while posicion < 20:

        dado = np.random.randint(1,7)

        posicion += dado

        if posicion > 20:
            posicion = 20

        if posicion in especiales:
            posicion = especiales[posicion]

        tiros += 1

    return tiros

# simulaciones
M = 1000000

resultados = [jugar() for _ in range(M)]

print("Promedio por simulación:")
print(np.mean(resultados))
```

Se simulan muchas partidas completas utilizando números aleatorios para representar las tiradas del dado.

El promedio de todas las simulaciones aproxima el valor esperado teórico.

Conclusión

El número promedio de tiradas puede calcularse:

analíticamente resolviendo un sistema lineal,
o experimentalmente mediante simulación Monte Carlo.

Ambos métodos producen resultados similares, validando el modelo probabilístico del juego.

#Juego del raton
Modelo del problema

El ratón se mueve aleatoriamente entre las casillas vecinas del tablero.

Los estados del sistema son:

$
\{0,1,2,3,4,5,6,7,8\}
$

donde:

- El estado \(7\) representa la comida (éxito).
- El estado \(8\) representa el shock (fracaso).

Estos dos estados son absorbentes.


Objetivo

Queremos encontrar:

$
P_i
$

donde \(P_i\) representa la probabilidad de llegar a la comida iniciando desde el estado \(i\).

En particular, buscamos:

$
P_0
$


 Relaciones entre estados

El ratón se mueve con igual probabilidad hacia cualquiera de las casillas vecinas.



Vecinos de cada estado

Estado 0

Desde el estado \(0\) el ratón puede ir a:

- \(1\)
- \(2\)

Entonces:

$
P_0=\frac{1}{2} P_1+\frac{1}{2} P_2
$



Estado 1

Los vecinos son:

$
0, 3, 7
$

Por lo tanto:

$
P_1=\frac{1}{3}(P_0+P_3+1)
$

ya que:

$
P_7=1
$

porque llegar a la comida significa éxito seguro.



Estado 2

Los vecinos son:

$
0,\ 3,\ 8
$

Entonces:

$
P_2=\frac{1}{3}(P_0+P_3+0)
$

debido a que:

$
P_8=0
$

porque llegar al shock significa fracaso seguro.

Estado 3

Los vecinos son:

$
1, 2, 4, 5
$

Por tanto:

$
P_3=\frac{1}{4}(P_1+P_2+P_4+P_5)
$

Estado 4

Los vecinos son:

$
3, 6, 7
$

Entonces:

$
P_4=\frac{1}{3}(P_3+P_6+1)
$

 Estado 5

Los vecinos son:

$
3, 6, 8
$

Por lo tanto:

$
P_5=\frac{1}{3}(P_3+P_6)
$

Estado 6

Los vecinos son:

$
4, 5
$

Entonces:

$
P_6=\frac{1}{2}(P_4+P_5)
$



```python
import numpy as np

# matriz del sistema
A = np.zeros((7,7))
b = np.zeros(7)

# P0
A[0,0] = 1
A[0,1] = -1/2
A[0,2] = -1/2

# P1
A[1,1] = 1
A[1,0] = -1/3
A[1,3] = -1/3
b[1] = 1/3

# P2
A[2,2] = 1
A[2,0] = -1/3
A[2,3] = -1/3

# P3
A[3,3] = 1
A[3,1] = -1/4
A[3,2] = -1/4
A[3,4] = -1/4
A[3,5] = -1/4

# P4
A[4,4] = 1
A[4,3] = -1/3
A[4,6] = -1/3
b[4] = 1/3

# P5
A[5,5] = 1
A[5,3] = -1/3
A[5,6] = -1/3

# P6
A[6,6] = 1
A[6,4] = -1/2
A[6,5] = -1/2

# resolver
P = np.linalg.solve(A,b)

print("Probabilidades de llegar a comida:\n")

for i,p in enumerate(P):
    print(f"P({i}) = {p:.4f}")

print("\nProbabilidad iniciando en 0:")
print(P[0])
```

Simulacion

```python
import numpy as np

vecinos = {
    0:[1,2],
    1:[0,3,7],
    2:[0,3,8],
    3:[1,2,4,5],
    4:[3,6,7],
    5:[3,6,8],
    6:[4,5]
}

def simulacion():

    estado = 0

    while estado not in [7,8]:

        estado = np.random.choice(vecinos[estado])

    return estado == 7

# número de simulaciones
N = 100000

resultados = [simulacion() for _ in range(N)]

print("Probabilidad aproximada:")
print(np.mean(resultados))
```

Método analítico

La probabilidad de alcanzar la comida satisface:

$
P_i=\sum_j p_{ij}P_j
$

con condiciones absorbentes:

$
P_7=1,\quad{P_8=0}
$

Esto genera un sistema de ecuaciones lineales.

Método de simulación

Se generan miles de trayectorias aleatorias del ratón hasta alcanzar:

- la comida (estado \(7\)),
- o el shock (estado \(8\)).

La proporción de veces que el ratón alcanza la comida aproxima la probabilidad real.

Conclusión

El problema puede resolverse:

- exactamente mediante ecuaciones de Markov,
- o aproximadamente mediante simulación Monte Carlo.

Ambos métodos producen resultados muy similares, validando el modelo probabilístico del sistema.
