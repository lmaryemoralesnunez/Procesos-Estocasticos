**Caminata Aleatoria 1D**


Para explicar mejor que es una caminata aleatoria en una dimension es un proceso estocastico donde un punto se mueve en pasos en donde van contandose uno por uno hacia la derecha o la izquierda.

Luego:
$$
X_{n+1} =
\begin{cases}
X_n + 1 & \text{con probabilidad } p \\
X_n - 1 & \text{con probabilidad } q
\end{cases}
$$
Teniendo que:
$$
p + q = 1
$$

Si $p = 0.5$ entonces la caminata es simétrica.  

Si $p > 0.5$ entonces existe un paso hacia la derecha.  

Si $p < 0.5$ entonces existe un paso hacia la izquierda.

Recordamos que el valor esperado se expresa como:
$$
E[\Delta X] = (1)(p) + (-1)(q)
$$

Sabemos que $q = 1 - p$:

$$
E[\Delta X] = p - (1 - p)
$$

$$
E[\Delta X] = 2p - 1
$$
Luego:
- Si $p = 0.5$, el valor esperado es $0$.
- Si $p > 0.5$, la caminata tiene tendencia positiva.
- Si $p < 0.5$, la caminata tiene tendencia negativa.

```python
import numpy as np
import matplotlib.pyplot as plt

def caminata_aleatoria_1D(n_ps, p, q):


    random = np.random.rand(n_ps)

    ps = [0]
    c = 0

    for i in random:
        if i < p:
            c += 1
        else:
            c -= 1
        ps.append(c)

    return ps


# Parámetros
p = float(input("Introduce el valor de p (entre 0 y 1): "))
q = 1 - p
n_ps = 100

# Generamos una sola caminata
trayectoria = caminata_aleatoria_1D(n_ps, p, q)

# Graficamos
plt.figure(figsize=(10,6))
plt.plot(trayectoria,color="#ff4da6", label="Caminata aleatoria")
plt.title(f'Caminata Aleatoria 1D (p = {p}, q = {q})')
plt.xlabel('Paso')
plt.ylabel('Posición')
plt.show()
```

**Conclusion**

Cuando $𝑝=0.5$, las trayectorias oscilan alrededor de cero.

Cuando $𝑝>0.5$, las trayectorias crecen.

Y cuando $𝑝<0.5$, decrecen.

La varianza aumenta con el número de pasos y aunque exista tendencia, las trayectorias individuales siguen siendo aleatorias, tambien el parámetro 𝑝 controla la dirección promedio del sistema.
